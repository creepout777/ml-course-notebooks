# Polynomial Regression

## Part A - Salary by position level

#### a - Importing libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#### b - Importing Dataset
The HR of a company has collected information from the internet (<b>e.g: glassdoor.com, indeed.com</b>) about salaries for different positions. 

The objective for this company is that, when carrying out interviews and receive <b>salary expectations</b> from the candidate, they want to investigate whether these expectations <b>are reasonable or not</b>, based on <b>Position</b> and <b>Level</b>. Note that the <b>Level</b> can be a continuous value (<b>e.g: 5.5</b>)

The company has hired you as a data scientist and Machine learning specialist to perform the previous task, based on  the <b>Polynomial Regression algorithm.</b> 


<b>Note:</b> In this notebook, we are going to compare the results of a Simple Linear Regression model with those of a <b>Polynomial Regression model</b>, and see their performance on a data that has a <b>non-linear relationship</b> between the Features and the Output

```python
# importing and displaying dataset 
data = pd.read_csv('../data/Position_Salaries.csv')
data
```

```python
# Creating Feature vector and output vector
x = data.iloc[:, 1:-1].values
y = data.iloc[:, -1].values
x
```

#### c - Training the Simple Linear Regression model

```python
from sklearn.linear_model import LinearRegression
l_reg = LinearRegression()
l_reg.fit(x, y)
```

#### d - Training the Polynomial Regression model

```python
from sklearn.preprocessing import PolynomialFeatures
# let's first create the polynomial features
pf = PolynomialFeatures(degree=4)
x_new = pf.fit_transform(x)
p_reg = LinearRegression()
p_reg.fit(x_new, y)
```

#### e - Visualizing Simple Linear Regression results

```python
plt.scatter(x, y, color='red')
plt.plot(x, l_reg.predict(x), color='blue')
plt.title('Simple Linear Regression')
plt.xlabel('Level')
plt.ylabel('Salary')
plt.show()
```

#### f - Visualizing Polynomial Regression results

```python
x_grid = np.arange(x.min(), x.max() + 0.1, 0.1).reshape(-1, 1)
plt.scatter(x, y, color='red')
plt.plot(x_grid, p_reg.predict(pf.transform(x_grid)), color='blue')
plt.title('Polynomial Regression (degree 4)')
plt.xlabel('Level')
plt.ylabel('Salary')
plt.show()
```

#### Try out different degree polynomial and the run the previous cell to see the change (e.g: Try degree = 3, then, degree = 4)

We use degree 4 here. Degree 2 underfits the top levels, degree 3 is close, and degree 4 passes near every point.

#### g- Performing prediction with Simple Linear Regression model 
Try predicting, what would be the salary for a Level of 5.5 with Simple Linear Regression model <b>(l_reg)</b>

```python
l_reg.predict([[5.5]])
```

#### e- Performing prediction with Polynomial Regression Regression model 
Try predicting, what would be the salary for a Level of 5.5 with <b>Polynomial Regression model (p_reg) </b>

```python
p_reg.predict(pf.transform([[5.5]]))
```

<b>Q:</b> What sounds more reasonable in this study, Simple Linear Regression prediction or Polynomial Regression prediction ???

**Answer:** the polynomial prediction (about 132,148) is the reasonable one. A level of 5.5 sits between *Country Manager* (110,000) and *Region Manager* (150,000), while the straight line predicts 249,500, more than a level 7 salary.

## Part B - Polynomial Regression on the power plant dataset
The same model on 9,568 readings with 4 features, evaluated on a held-out test set (R²).

```python
dataset = pd.read_csv('../data/Power_Plant.csv')
X = dataset.iloc[:, :-1].values
Y = dataset.iloc[:, -1].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=0)
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
poly_reg = PolynomialFeatures(degree=4)
regressor = LinearRegression()
regressor.fit(poly_reg.fit_transform(X_train), y_train)
y_pred = regressor.predict(poly_reg.transform(X_test))
```

```python
from sklearn.metrics import r2_score, mean_squared_error
np.set_printoptions(precision=2)
print(np.concatenate((y_pred.reshape(-1, 1), np.asarray(y_test).reshape(-1, 1)), 1)[:10])
print('R2  :', round(r2_score(y_test, y_pred), 4))
print('RMSE:', round(float(np.sqrt(mean_squared_error(y_test, y_pred))), 3), 'MW')
```
