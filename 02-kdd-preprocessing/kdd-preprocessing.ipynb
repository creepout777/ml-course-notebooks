# Data Preprocessing
Data preprocessing is a crucial step in machine learning that involves transforming raw data into a clean, structured, and meaningful format for better model performance. The main purposes of data preprocessing include:
1) <b>Handling Missing Data</b>: Filling in missing values using imputation techniques (mean, median, mode) or removing incomplete records to prevent biased learning.
2) <b>Removing Noise and Outliers</b>: Identifying and handling anomalies that could distort the learning process.
3) <b>Feature Scaling</b>: Normalizing or standardizing numerical features to ensure all variables contribute equally to the model, especially for distance-based algorithms 
4) <b>Data Transformation</b>: Converting categorical data into numerical formats using encoding techniques (e.g., one-hot encoding, label encoding)
5) <b>Enhancing Model Performance</b>: Well-preprocessed data leads to better generalization, higher accuracy, and more efficient training.

## 2 - Importing Dataset
KDDCup99 is a benchmark dataset used for evaluating intrusion detection systems (IDS) including machine learning-based. It was created as part of the KDD Cup 1999, an international data mining competition organized by the Knowledge Discovery and Data Mining (KDD) group. The dataset is derived from the 1998 DARPA Intrusion Detection Evaluation Program, which simulated network traffic over a military network.

### Dataset Description:
KDDCup99 consists of simulated network traffic data and includes both normal connections and various types of cyber attacks. Each connection is represented by 41 features, categorized into:
1) Basic features: Packet and connection attributes (e.g., protocol type, duration)
2) Content features: Data inside the packet (e.g., number of failed login attempts)
3) Traffic features:  Statistical information over time windows (e.g., number of connections from the same IP in the last 2 seconds).

### Attack Categories:
Attacks in KDDCup99 are grouped into four main types:
1) Denial-of-Service (DoS): Overwhelming network resources (e.g., smurf, teardrop, pod).
2) Probe – Scanning vulnerabilities in the network (e.g., nmap, satan, ipsweep).
3) User-to-Root (U2R) – Gaining root access by exploiting system vulnerabilities (e.g., buffer_overflow, rootkit).
4) Remote-to-Local (R2L) – Gaining unauthorized local access (e.g., phf, ftp_write, guess_passwd).

## 3 - Encoding Categorical Data
The vast majority of ML models expect features as numerical values. Thus, categorical variables/features (Like protocol_type in our case), should be converted to  good representative numerical values.

### 3-1 :  Encoding Categorical features (protocol_type, service, flag)

The representative numerical values shouldn't establish any order between the categorical data. This will be done using<b>One Hot Encoding</b>.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(sparse_output=False), [1, 2, 3])], remainder='passthrough')
x = ct.fit_transform(x).astype(float)   # all values are numbers now
print(x)
```

<b>Q1: After the previous step, how much feature are now in x ?</b>

**Answer:** one-hot encoding turns the 3 categorical columns into 3 + 66 + 11 = 80 columns, so `x` has 80 + 38 = **118 features**.

## 5 - Feature Scaling 
Feature scaling is the process of transforming numerical variables into a specific range or distribution to ensure that all features contribute equally to a machine learning model. It is especially important for algorithms that rely on distance calculations or gradient-based optimization.

Feature scaling provides the following characteristics:
1) <b>Prevents Dominance of Large-Scale Features</b>: Features with larger numerical ranges may predominate smaller-scale features, leading to biased results
2) <b>Improves Model Convergence</b>: Some ML algorithms converge faster when features are on a similar scale

### 5-1 Normalization Vs. Standardization
Normalization and  Standardization are 2 of many techniques that performs Feature scaling, Following is the mathematical foundation of each:

<b>Normalization</b>: Convert the values approximatively in the range <b>[0,1]</b>. <b>This is recommended when we have a normal distribution of the values</b>.


$$
x_{norm} = \frac{x - min(x)}{max(x) - min(x)}
$$

<b>Standardization</b>: Convert the values approximatively in the range <b>[-3,3]</b>. <b>Useful in most of the cases, even when not having a normal distribution</b>

$$
x_{stand} = \frac{x - mean(x)}{std(x)}
$$

Overall, since <b>Standardization</b> is <b>Useful in most of the cases</b>, we're gonna go for it.

### Note !!!!! 
<b>Standardization should be applied to numerical values.</b>

With `remainder='passthrough'`, the one-hot columns come first, so the numerical values start at index 80 (computed below as `x_train.shape[1] - 38`).. From indices 1 to 80, there are only categorical values
