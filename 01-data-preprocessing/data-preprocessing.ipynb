# Data Preprocessing
Data preprocessing is a crucial step in machine learning that involves transforming raw data into a clean, structured, and meaningful format for better model performance. The main purposes of data preprocessing include:
1) <b>Handling Missing Data</b>: Filling in missing values using imputation techniques (mean, median, mode) or removing incomplete records to prevent biased learning.
2) <b>Removing Noise and Outliers</b>: Identifying and handling anomalies that could distort the learning process.
3) <b>Feature Scaling</b>: Normalizing or standardizing numerical features to ensure all variables contribute equally to the model, especially for distance-based algorithms 
4) <b>Data Transformation</b>: Converting categorical data into numerical formats using encoding techniques (e.g., one-hot encoding, label encoding)
5) <b>Enhancing Model Performance</b>: Well-preprocessed data leads to better generalization, higher accuracy, and more efficient training.

## 1 - Importing libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2 - Importing Dataset
This is a dataset regrouping information about a company that sells a specific product over 3 countries: <b> France, Spain and Germany </b>. They collected data about their customers, whether they purchased the product of not.

For a customer, <b> residing in one of these countries, having an age, and a salary </b>, the company will want to know in the future whether that customer will purchase the product or not, using <b> Machine Learning .

The company has hired you as a data scientist and Machine learning specialist to perform the previous task. But before that, <b>Pre-processing</b> should be done conveniently

```python
# importing and displaying dataset 
data =pd.read_csv("../data/Data_Preprocessing.csv")
data
```

```python
# Creating Feature vector and output vector
x = data.iloc[:,:3].values
y = data.iloc[:,3].values
x
```

## 3 - Handling missing Data
* By deleting rows having missing column data
* By replacing the missing information with : 1-The mean. 2-The median. 3-The most frequent

```python
from sklearn.impute import SimpleImputer 
myImputer = SimpleImputer(missing_values = np.nan, strategy = 'median')
myImputer.fit(x[:, 1:])
x[:, 1:] = myImputer.transform(x[:, 1:])
x
```

#### What are your observations on the results ?

**Observation:** the missing *Age* and *Salary* values were replaced by the median of each column.

## 3 - Encoding Categorical Data
The vast majority of ML models expect features as numerical values. Thus, categorical variables/features (Like country in our case), should be converted to  good representative numerical values.

### 3-1 :  Encoding Categorical features (Country)

The representative numerical values shouldn't establish any order between the categorical data. Here we have France, Germany and Spain, and it would be inappropriate to encode them by assigning an integer <b> 0, 1, 2 </b>to each of these countries, since it's like forcing some kind of order between them, they reprensent a similar concept.

The solution for that is what we call <b>One Hot Encoding</b> as shown below.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(transformers = [('encoder', OneHotEncoder(), [0])], remainder = 'passthrough')
x = ct.fit_transform(x)
x
```

### What are your observations on the results ?

**Observation:** column 0 (*Country*) became three binary columns, one per country, placed first in `x`.

### 3-2 :  Encoding the output (Purchased)
When we have only to categories (in this case Yes/No), we can use a more memory efficient way based on <b>LabelEncoder</b> instead of OneHotEncoder.

Indeed, using it with only two values:
1) There will be no ordinal misinterpretation.
2) It reduces dimensionality compared to One-Hot Encoding

**Note:** `LabelEncoder` suits targets and ordered or binary values such as Yes/No.

**Note:** `LabelEncoder` maps each label to one integer, while `OneHotEncoder` creates one binary column per category.

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
y = encoder.fit_transform(y)
y
```

## 4 - Splitting Data into Training and Test Sets
In Machine Learning, generally we divide our dataset into 2 separate sets: Training set and Test set.

* <b>Training set</b>: is a portion of data on which we are training our ML model using some ML algorithm.
* <b>Test set</b>: This is supposed to be a brand new data on which we test the performance of the ML model, by comparing the predicted values (from the model), with the actual ones.

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=1)

x_train
```

## 5 - Feature Scaling 
Feature scaling is the process of transforming numerical variables into a specific range or distribution to ensure that all features contribute equally to a machine learning model. It is especially important for algorithms that rely on distance calculations or gradient-based optimization.

Feature scaling provides the following characteristics:
1) <b>Prevents Dominance of Large-Scale Features</b>: Features with larger numerical ranges may predominate smaller-scale features, leading to biased results
2) <b>Improves Model Convergence</b>: Some ML algorithms converge faster when features are on a similar scale

### 5-1 Normalization Vs. Standardization
Normalization and  Standardization are 2 of many techniques that performs Feature scaling, Following is the mathematical foundation of each:

<b>Normalization</b>: Convert the values approximatively in the range <b>[0,1]</b>. <b>This is recommended when we have a normal distribution of the values</b>.


$$
x_{norm} = \frac{x - min(x)}{max(x) - min(x)}
$$

<b>Standardization</b>: Convert the values approximatively in the range <b>[-3,3]</b>. <b>Useful in most of the cases, even when not having a normal distribution</b>

$$
x_{stand} = \frac{x - mean(x)}{std(x)}
$$

Overall, since <b>Standardization</b> is <b>Useful in most of the cases</b>, we're gonna go for it.

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
x_train[:, 3:] = sc.fit_transform(x_train[:, 3:])
x_test[:, 3:] = sc.transform(x_test[:, 3:])
x_train
```
