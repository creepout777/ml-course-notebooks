# Classification

Classification in Machine Learning (ML) is a type of supervised learning where the goal is to predict a category or class label for a given input based on previously labeled data.


### A) Types of Classification:

<b>Binary Classification:</b> Two classes only (e.g., "yes" or "no").
<b>Multi-class Classification:</b> More than two classes (e.g., classifying animals: cat, dog, horse).

### B) Algorithms:
<b>1) Logistic Regression:</b>
* Logistic Regression is a supervised learning algorithm used for classification tasks, not regression 
* The goal is to estimate the probability that a data point belongs to a particular class, usually class 1 (positive class).
* It’s based on a linear model, but applies a sigmoid function to the output to convert it into a probability between 0 and 1:

$$
z = W_{1}x_{1}+W_{2}x_{2} + W_{3}x_{3} +...+ W_{n}x_{n} + b
$$
$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$


Then:

If $\sigma(z)$ ≥ 0.5, predict class 1

If $\sigma(z)$ < 0.5, predict class 0

Dataset: 400 users of a social network with their **Age** and **Estimated Salary**, and whether they bought a product after seeing an ad (**Purchased**).

## Importing the libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Importing the dataset

```python
dataset = pd.read_csv('../data/Social_Network_Ads.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

## Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python
print(X_train[:10])
```

```python
print(y_train[:10])
```

```python
print(X_test[:10])
```

```python
print(y_test[:10])
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
print(X_train[:10])
```

```python
print(X_test[:10])
```

## Training the Logistic Regression model on the Training set

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state=0)
classifier.fit(X_train, y_train)
```

## Predicting a new result
Try predict the outcome for an age of 30 and a salary of 87000

```python
print(classifier.predict(sc.transform([[30, 87000]])))
```

## Predicting the Test set results

```python
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1)[:15])
```

## Making the Confusion Matrix + Computing Accuracy

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```

**Result:** 89 of the 100 test users are classified correctly (65 true negatives, 24 true positives, 3 false positives, 8 false negatives). The decision boundary is a straight line, which is why some buyers near it are missed.

## Visualising the Training set results

```python
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_train), y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 0.25),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 250))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], color = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Logistic Regression (Training set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
```

## Visualising the Test set results

```python
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_test), y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 0.25),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 250))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], color = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Logistic Regression (Test set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
```
